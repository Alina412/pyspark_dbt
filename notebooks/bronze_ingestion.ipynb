## SPARK STREAMING

```python
source_tables = ["customers", "drivers", "locations", "payments", "trips", "vehicles"]
```

```python
for tbl in source_tables: 
    df_batch = spark.read.format("csv").option("header", True).option("inferSchema", True).load(f"/Volumes/pyspark_dbt/source/source_data/{tbl}")
    schema_tbl = df_batch.schema

    df = spark.readStream.format("csv").option("header", True).schema(schema_tbl).load(f"/Volumes/pyspark_dbt/source/source_data/{tbl}")

    df.writeStream.format("delta").outputMode("append").option("checkpointLocation", f"/Volumes/pyspark_dbt/bronze/checkpoint/{tbl}")\
        .trigger(once = True).toTable(f"pyspark_dbt.bronze.{tbl}")
```
